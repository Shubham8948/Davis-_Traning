```
# create a function to calculate simple intrest

def calculate_si(p, r, t):
    si = (p * r * t) / 100
    return si

P = float(input("Enter P: "))
R = float(input("Enter R: "))
T = float(input("Enter T: "))

result = calculate_si(P, R, T)

print("Simple Interest =", result)
```
