```
 # to calculaate the electricity bill based on units consumed
units = int(input("Enter units: "))

bill = 0

if units > 150:
    bill = (50 * 4) + (100 * 6) + (units - 150) * 8
elif units > 50:
    bill = (50 * 4) + (units - 50) * 6
else:
    bill = units * 4

print("Total bill amount =", bill)
```
