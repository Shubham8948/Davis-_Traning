```
simple_interest = lambda p, r, t: (p * r * t) / 100

P = float(input("Enter P: "))
R = float(input("Enter R: "))
T = float(input("Enter T: "))

SI = simple_interest(P, R, T)

print("Simple Interest =", SI)
```
